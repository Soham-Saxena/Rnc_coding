```python
class Node():
    def __init__(self, value):
        self.data = value
        self.ref = None
    def reference(self, ref):
        self.ref = ref
    def setval(self, value):
        self.data = value
    
```

```python
class Linked_list():
    def __init__(self):
        self.head = None
        self.nodes = 0
    def add_end(self, element):
        temp = self.head
        self.nodes += 1
        print(f"Number of nodes in the list: {self.nodes}")
        element = Node(element)
        if self.head is None:
            self.head = element
        else:
            temp = self.head
            i = 0
            while i < self.nodes -1:
                if i == self.nodes - 2:
                    temp.ref = element
                temp = temp.ref
                i += 1
            temp.ref = self.head
    def add_begining(self, element):
        temp = self.head
        self.nodes += 1
        self.head = Node(element)
        if temp is not None:
            self.head.ref = temp
            temp = self.head
            i = 0
            while i < self.nodes - 1:
                temp = temp.ref
                i += 1
            temp.ref = self.head
    def traverse_list(self):
        temp = self.head
        if temp is None:
            print("Linked list is empty.")
        else:
            for i in range(self.nodes):
                print(temp.data, end = " ")
                temp = temp.ref
            print("")
    def traverse_from(self, node):
        temp = self.head
        i = 1
        while i < node:
            temp = temp.ref
            i += 1
        i = 0
        print(f"Linked List contains (starting from node {node}): ", end = '')
        while i < self.nodes:
            print(temp.data, end = " ")
            temp = temp.ref
            i += 1
    def search(self, element):
        nodes = []
        c_node = 1
        temp = self.head
        if temp is None:
            print("Linked list is empty.")
        else:
            for i in range(self.nodes):
                if temp.data == element:
                    nodes.append(c_node)
                c_node += 1
                temp = temp.ref
        if (len(nodes) == 1):
            return nodes[0]
        elif (len(nodes) > 1):
            return nodes
        else:
            return -1
    def delete_node(self, node):
        temp1 = self.head
        if temp1 is None:
            print("Linked list is empty")
        elif (node - 2 < 0):
            self.head = temp1.ref
            for i in range(self.nodes - 1):
                temp1 = temp1.ref
            temp1.ref = self.head
            self.nodes -= 1
        elif (node == self.nodes):
            for i in range(self.nodes - 2):
                temp1 = temp1.ref
            temp1.ref = self.head
            self.nodes -= 1
        else:
            for i in range(node-2):
                temp1 = temp1.ref
            temp2 = temp1.ref.ref
            temp1.ref = temp2
            self.nodes -= 1
    def delete_element(self, element):
        count = 0
        node = self.search(element)
        if node == -1:
            print("Element not found in list.")
        else:
            if isinstance(node, int):
                self.delete_node(node)
                print("Element deleted succesfully.")
            else:
                print(f"{len(node)} occurances of elements found. What would you like to do?")
                choice = int(input("Delete all (1)\nDelete specific occurances (2)"))
                while choice in range(1, 2):
                    print("Invalid input.")
                    choice = int(input("Delete all (1)\nDelete specific occurances (2)"))
                if choice == 1:
                    for n in node:
                        self.delete_node(n- count)
                        count += 1
                    print("All occurances deleted succesfully.")
                else:
                    print(f"Elements occur at nodes: {node}")
                    nodes = list(map(int, input("Enter the nodes u want to delete: ").split()))
                    print(nodes)
                    while not (set(nodes).issubset(set(node))):
                        print("Nodes specified not in range.")
                        nodes = list(map(int, input("Enter the nodes u want to delete: ").split()))
                    for n in nodes:
                        self.delete_node(n - count)
                        count += 1
                    print("Selected nodes deleted.")

                    
```
