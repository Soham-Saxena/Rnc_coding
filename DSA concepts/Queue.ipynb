```python
def enqueue(List, element):
    List.append(element)
    print('Element added succesfully.')
```

```python
def dequeue(List):
    a = List.pop(0)
    print(f"Element {a} removed succesfully.")
```

```python
def display(List):
    print(f"Queue in use: {List}")
```

```python
List = []
while True:
    choice = input("Enter operation to perform on Queue:\nEnqueue(1)\nDequeue(2)\nDisplay(3)\nExit(4)")
    try:
        choice = int(choice)
    except:
        print("Input entered is not an integer")
        continue
    if choice == 1:
        element = input("Enter element to enqueue: ")
        enqueue(List, element)
    elif choice == 2:
        dequeue(List)
    elif choice == 3:
        display(List)
    elif choice == 4:
        print("Exiting")
        break
    else:
        print("Invalid input, try again")
```
