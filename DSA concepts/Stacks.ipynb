```python
def push(List, element):
    List.append(element)
```

```python
def pop(List):
    List.pop()
```

```python
def peek(List):
    print(f"Element on top of stack: {List[-1]}")
```

```python
def display(List):
    print(f"Current Stack: {List}")
```

```python
def check(List):
    if not len(List):
        print("Stack is empty.")
    print("Stack is not empty.")
```

```python
List = []
while True:
    choice = input("Enter operations to perform on stack:\nPush(1)\nPop(2)\nPeek(3)\nDisplay(4)\nCheck(5)\nExit(6)")
    try:
        int(choice)
    except:
        print("Input given is not an integer")
        continue
    if int(choice) == 1:
        element = int(input("Enter element to push: "))
        push(List, element)
        print("Element pushed succesfully")
    elif int(choice) == 2:
        if len(List):
            pop(List)
            print("ELement on top popped successfully.")
        else:
            print("Stack is empty.")
    elif int(choice) == 3:
        if len(List):
            peek(List)
        else:
            print("Stack is empty.")
    elif int(choice) == 4:
        display(List)
    elif int(choice) == 5:
        check(List)
    elif int(choice) == 6:
        print('Exiting')
        break
    else:
        print("Invalid input, try again.")
```
